```python
from pyspark.sql import functions as f

base_path = "/Volumes/electricity_catalog/electricity_schema/raw/"

karnataka_weather = spark.read.format("csv").\
    option("inferSchema","true").\
    option("header","true").\
    load(base_path+"Karnataka.csv")

maharashtra_weather = spark.read.format("csv").\
    option("inferSchema","true").\
    option("header","true").\
    load(base_path+"Maharashtra.csv")

energy_raw = spark.read.format("csv").\
    option("inferSchema","true").\
    option("header","true").\
    load(base_path+"corrected_daily_energy_met_MU.csv")
peak_raw = spark.read.format("csv").\
    option("inferSchema","true").\
    option("header","true").\
    load(base_path+"corrected_peak_met_MW.csv")

karnataka_weather.limit(10).display()

maharashtra_weather.limit(10).display()

energy_raw.limit(10).display()

peak_raw.limit(10).display()
```

```python
karnataka_weather.printSchema()
maharashtra_weather.printSchema()
energy_raw.printSchema()
peak_raw.printSchema()
```

```python
peak_clean = peak_raw

for old_col in peak_clean.columns:
    new_col = (
        old_col
        .strip()
        .replace(" ", "_")
        .replace(",", "_")
        .replace(";", "_")
        .replace("{", "_")
        .replace("}", "_")
        .replace("(", "_")
        .replace(")", "_")
        .replace("\n", "_")
        .replace("\t", "_")
        .replace("=", "_")
    )

    peak_clean = peak_clean.withColumnRenamed(
        old_col,
        new_col
    )

peak_clean = peak_clean.withColumnRenamed(
    "Date",
    "date"
)
```

```python
#Writing peak data to table
peak_clean.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "electricity_catalog.electricity_schema.bronze_peak"
    )
```

```python
energy_clean = energy_raw

for old_col in energy_clean.columns:
    new_col = (
        old_col
        .strip()
        .replace(" ", "_")
        .replace(",", "_")
        .replace(";", "_")
        .replace("{", "_")
        .replace("}", "_")
        .replace("(", "_")
        .replace(")", "_")
        .replace("\n", "_")
        .replace("\t", "_")
        .replace("=", "_")
    )

    energy_clean = energy_clean.withColumnRenamed(
        old_col,
        new_col
    )

```

```python
#Writing energy data to table
energy_clean.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "electricity_catalog.electricity_schema.bronze_energy"
    )
```

```python
#Writing karnataka weather data to table

karnataka_weather.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "electricity_catalog.electricity_schema.bronze_karnataka_weather"
    )
```

```python
maharashtra_weather.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "electricity_catalog.electricity_schema.bronze_maharashtra_weather"
    )
```
